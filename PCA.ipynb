```python
from sklearn.decomposition import PCA
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
scaled_dataset = sc.fit_transform(df)
```

```python
scaled_dataset
```

```python
pca = PCA()
```

```python
cov_mat = np.cov(scaled_dataset.T)
```

```python
cov_mat.shape
```

```python
eig_val, eig_vector = np.linalg.eig(cov_mat)
```

```python
tot = sum(eig_val)
```

```python
var_exp = [(i/tot)*100 for i in sorted(eig_val, reverse = True)]
```

```python
var_exp
```

```python
cum_var_exp = np.cumsum(var_exp)
```

```python
cum_var_exp
```

```python
df_cum = pd.DataFrame(cum_var_exp)
```

```python
df_cum.head(25)
```

```python
pca = PCA(n_components =25, random_state = 20)
```

```python
pca.fit(scaled_dataset)
```

```python
pca_x = pca.fit_transform(scaled_dataset)
```

```python
pca_x
```

```python
pca_x = pd.DataFrame(pca_x)
```

```python
pca_x 
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(pca_x, target, test_size = 0.3, random_state = 20 )
```

```python
x_train
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
df_model = RandomForestClassifier()
```

```python
df_model.fit(x_train, y_train)
```

```python
pred = df_model.predict(x_test)
```

```python
from sklearn.metrics import *
```

```python
accuracy = accuracy_score(y_test, pred)
```
